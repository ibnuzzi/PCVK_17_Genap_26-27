```python
#6. Membuat bentuk Geometri 2D dari OpenCV. Diawali dengan pembuatan black image dengan tipe data int16.

import numpy as np
black_img = np.zeros(shape=(512,512,3),dtype=np.int16)
plt.imshow(black_img)
```

```python
#Menambahkan Persegi Panjang Pertama (Hijau)

import cv2 as cv

#perhatikan koordinat titik2 pt1 dan pt2
cv.rectangle(black_img,pt1=(384,0),pt2=(510,150),color=(0,255,0),thickness=10)
plt.imshow(black_img)
```

```python
#Menambahkan Persegi Panjang Kedua (Biru)
cv.rectangle(black_img,pt1=(200,200),pt2=(300,300),color=(0,0,255),thickness=15)
plt.imshow(black_img)
```

```python
#Menambahkan Lingkaran (Merah)
cv.circle(black_img,center=(100,100),radius=50,color=(255,0,0),thickness=8)
plt.imshow(black_img)
```

```python
#Menambahkan Garis (Putih)
cv.line(black_img,pt1=(0,0),pt2=(512,512),color=(255,255,255),thickness=5)
plt.imshow(black_img)
```

```python
#Menambahkan Teks
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img,text='Hello',org=(10,500),fontFace=font,fontScale=4,color=(255,255,0),thickness=2,lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
#Membuat Black Image Kedua (int32) dan Inisialisasi Array
black_img2=np.zeros(shape=(512,512,3),dtype=np.int32)
plt.imshow(black_img2)

vertices = np.array([[100,300],[200,200],[400,300],[200,400]],dtype=np.int32)
vertices

pts = vertices.reshape((-1,1,2)) # nilai 2 untuk menunjukkan bahwa tiap titik dibuat 3 channel yg mewakili R, G, dan B
pts
```

```python
#Menambahkan Polyline pada Black Image Kedua
cv.polylines(black_img2,[pts],isClosed=True,color=(255,0,0),thickness=5)
plt.imshow(black_img2)
```

```python
#Kode Soal 2: Channel Red-Blue & Green-Blue

# Tampilan Red-Blue (Menonaktifkan channel Green di indeks 1)
img_rb = imgRGB.copy()
img_rb[:, :, 1] = 0
plt.figure(figsize=(10,5))
plt.subplot(1,2,1); plt.imshow(img_rb); plt.title("Red-Blue")

# Tampilan Green-Blue (Menonaktifkan channel Red di indeks 0)
img_gb = imgRGB.copy()
img_gb[:, :, 0] = 0
plt.subplot(1,2,2); plt.imshow(img_gb); plt.title("Green-Blue")
plt.show()
```

```python
#Kode Soal 3: Menampilkan Baris 20-115, Kolom 25-120

img_crop = imgRGB[20:115, 25:120]
plt.imshow(img_crop)
plt.title("Slicing Baris & Kolom")
plt.show()
```

```python
#Kode Soal 4: Menampilkan Baris 5-30, Semua Kolom, Channel Red Saja

img_red_slice = imgRGB.copy()
img_red_slice = img_red_slice[5:30, :] # Ambil baris 5-30, semua kolom
img_red_slice[:, :, 1] = 0 # Nol-kan nilai Green
img_red_slice[:, :, 2] = 0 # Nol-kan nilai Blue
plt.imshow(img_red_slice)
plt.title("Baris 5-30, Semua Kolom, Channel Red")
plt.show()
```

```python
#Kode Soal 5: Membuat 5 Kotak Ukuran & Warna Acak

import random

img_kotak = imgRGB.copy()
tinggi, lebar, _ = img_kotak.shape

for _ in range(5):
    # Membuat koordinat acak
    x1, y1 = random.randint(0, lebar//2), random.randint(0, tinggi//2)
    x2, y2 = random.randint(lebar//2, lebar), random.randint(tinggi//2, tinggi)

    # Membuat warna RGB acak
    warna_acak = (random.randint(0,255), random.randint(0,255), random.randint(0,255))

    cv2.rectangle(img_kotak, (x1, y1), (x2, y2), warna_acak, thickness=3)

plt.imshow(img_kotak)
plt.title("5 Kotak Acak")
plt.show()
```

```python
#Kode Soal 6: Gambar Terbalik

# Parameter -1 membalikkan gambar secara vertikal sekaligus horizontal
img_terbalik = cv2.flip(imgRGB, -1)
plt.imshow(img_terbalik)
plt.title("Gambar Terbalik")
plt.show()
```

```python
#Kode Soal 7: Bentuk Rectangle & Circle pada Foto Sendiri

# Silakan sesuaikan nama file dengan foto kegiatan yang Anda unggah
img_foto = cv2.imread("/content/drive/MyDrive/PCVK/foto_kegiatan_saya.jpg")
img_foto_rgb = cv2.cvtColor(img_foto, cv2.COLOR_BGR2RGB)

# Mendapatkan dimensi tinggi dan lebar gambar asli
h, w, _ = img_foto_rgb.shape

# Menentukan titik koordinat tengah wajah (estimasi: tengah secara horizontal, 37% dari atas)
titik_x_tengah = int(w * 0.49)
titik_y_tengah = int(h * 0.37)

# Menentukan ukuran lebar dan tinggi area wajah
lebar_wajah = int(w * 0.15)
tinggi_wajah = int(h * 0.11)

# Menentukan titik pt1 (kiri atas) dan pt2 (kanan bawah) untuk rectangle
pt1 = (titik_x_tengah - int(lebar_wajah/2), titik_y_tengah - int(tinggi_wajah/2))
pt2 = (titik_x_tengah + int(lebar_wajah/2), titik_y_tengah + int(tinggi_wajah/2))

# Membuat Rectangle (Hijau) dan Circle (Merah)
cv2.rectangle(img_foto_rgb, pt1=pt1, pt2=pt2, color=(0, 255, 0), thickness=10)
cv2.circle(img_foto_rgb, center=(titik_x_tengah, titik_y_tengah), radius=int(w * 0.08), color=(255, 0, 0), thickness=10)

plt.imshow(img_foto_rgb)
plt.title("Rectangle dan Circle pada Wajah")
plt.show()
```

```python
#Kode Soal 8 & 9: Rectangle Sudut Kiri Bawah & Teks Nama File

# Asumsi menggunakan gambar 'kitten'
img_kucing = cv2.imread("/content/drive/MyDrive/PCVK/kucing.jpg")
img_kucing_rgb = cv2.cvtColor(img_kucing, cv2.COLOR_BGR2RGB)

# Mendapatkan dimensi tinggi dan lebar gambar
h, w, _ = img_kucing_rgb.shape

# Menentukan proporsi ukuran kotak (misal: lebar 35% dan tinggi 15% dari gambar asli)
lebar_kotak = int(w * 0.35)
tinggi_kotak = int(h * 0.15)

# Soal 8: Membuat Rectangle di sudut kiri bawah (Warna Biru RGB: 0, 0, 255)
cv2.rectangle(img_kucing_rgb, pt1=(0, h - tinggi_kotak), pt2=(lebar_kotak, h), color=(0, 0, 255), thickness=10)

# Soal 9: Menambahkan teks nama file di area kotak tersebut
font = cv2.FONT_HERSHEY_SIMPLEX
# Skala font disesuaikan dengan tinggi gambar agar tidak terlalu kecil atau besar
skala_font = h * 0.002
cv2.putText(img_kucing_rgb, text='kucing.jpg', org=(30, h - int(tinggi_kotak/2.5)), fontFace=font, fontScale=skala_font, color=(255, 255, 0), thickness=4, lineType=cv2.LINE_AA)

plt.imshow(img_kucing_rgb)
plt.title("Rectangle dan Teks pada Kucing")
plt.show()
```
